# Updated Initial Observations
The dataset contains 1,067,371 transaction records across 8 columns.
Each row represents a transaction (or purchased item) rather than a unique customer. Customer-level features will need to be engineered before clustering.
The dataset spans from December 2009 to December 2011, providing approximately two years of transaction history.
The Customer ID column contains 243,007 missing values, which will require investigation during data cleaning.
The Description column contains only 4,382 missing values, while the remaining columns have no missing values.
The summary statistics show negative values in both Quantity and Price, indicating that there are special transactions (such as returns, cancellations, or corrections) that should be investigated before modeling.
The dataset contains 53,628 unique invoices, 5,305 unique stock codes, and transactions from 43 different countries, with the majority originating from the United Kingdom.
