# Cleaning data and removing rows


```python
import pandas as pd 
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT/"data"/"raw"/"online_retail_II.xlsx"
```

```python
df_2010 = pd.read_excel(DATA_PATH, sheet_name = "Year 2009-2010")
df_2011 = pd.read_excel(DATA_PATH, sheet_name = "Year 2010-2011")
df = pd.concat([df_2010, df_2011], ignore_index=True)
```

```python
df.shape
```

removing rows with cancellations first 

```python
df["Invoice"] = df["Invoice"].astype(str)
df = df[~df["Invoice"].str.startswith("C")]
df.shape
```

removing customer ids

```python
df = df[df["Customer ID"].notna()]
df.shape
```

removing negative prices

```python
df = df[df["Price"] >= 0]
df.shape
```

removing duplicates

```python
df = df.drop_duplicates()
df.shape
```
